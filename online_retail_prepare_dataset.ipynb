```python
import pandas as pd

retail = pd.read_csv(
    "C:/Users/maazk/Downloads/online_retail_II.csv",
    encoding="latin1", dtype={"Invoice": str, "StockCode": str})


retail.head()
```

```python
# Convert InvoiceDate to ISO format
retail["InvoiceDate"] = pd.to_datetime(
    retail["InvoiceDate"],
    format="%m/%d/%y %H:%M"
).dt.strftime("%Y-%m-%d %H:%M:%S")

# Check
print(retail[["InvoiceDate"]].head())
```

```python
# 4. Convert Customer_ID to nullable integer
retail["Customer_ID"] = retail["Customer ID"].astype("Int64")

# 5. Check Customer_ID
print(retail["Customer_ID"].head())
```

```python
# 6. Save the cleaned data
retail.to_csv(
    "C:/Users/maazk/Downloads/online_retail_cleaned.csv",
    index=False
)
```
